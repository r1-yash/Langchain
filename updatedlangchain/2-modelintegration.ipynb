## Streaming and Batch

### Streaming
Models can stream their output in real tme while the content is being generated. It improves user experience. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time.

```python
for chunk in model.stream("teach me about streaming in langchain in short"):
    print(chunk.text)
```

```python
responses = model.batch([
    "What is a red",
    "what is batch in langchain",
    "did spain win world cup or argentina"
])
responses
```
